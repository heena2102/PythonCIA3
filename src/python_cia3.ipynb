```
class Vehicle:
    """Base class demonstrating Encapsulation and Abstraction"""
    def __init__(self, brand, model):
        self.brand = brand  # Public attribute
        self.model = model
        self.__engine_status = False  # Private attribute (Encapsulation)

    def start_engine(self):
        self.__engine_status = True
        print(f"{self.brand} {self.model} engine started.")

    def stop_engine(self):
        self.__engine_status = False
        print(f"{self.brand} {self.model} engine stopped.")

    def drive(self):
        # Method to be overridden (Polymorphism)
        print("The vehicle is moving.")

class Car(Vehicle):
    """Child class demonstrating Inheritance"""
    def __init__(self, brand, model, doors):
        # Use super() to call the constructor of the parent class
        super().__init__(brand, model)
        self.doors = doors

    def drive(self):
        # Overriding the parent method (Polymorphism)
        print(f"The {self.brand} {self.model} car is driving smoothly on {self.doors} doors.")

    def open_trunk(self):
        print("Trunk opened.")

# Instantiate and demonstrate OOP
my_car = Car("Tesla", "Model 3", 4)
my_car.start_engine()
my_car.drive()
my_car.stop_engine()
```
